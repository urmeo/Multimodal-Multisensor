# Group Hrv

Group-level HRV (SDNN) across sessions.

**Reads:** `data/group_results/ (10-participant summaries)`  
**Shared code:** `import mms` (loaders in `mms.io`, metrics in `mms.hrv` / `mms.stats`).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

OUTPUT = '../data/group_results'

# load hrv data
results_df = pd.read_csv(f'{OUTPUT}/HRV_SDNN.csv')

print(results_df)

# plot standard deviations
ax = results_df.set_index('Participant').plot(figsize=(15, 8), marker='o')

plt.title('Standard Deviation of HRV (SDNN) for Each Participant Across Three Sessions')
plt.ylabel('Standard Deviation of HRV (SDNN) (ms)')
plt.xlabel('Participant')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# reference lines
ax.axhline(y=50, color='blue', linestyle=':', label='Normal SDNN')
ax.axhline(y=100, color='purple', linestyle=':', label='Moderate SDNN')
ax.axhline(y=150, color='green', linestyle=':', label='High SDNN')

# merge legend handles
handles, labels = ax.get_legend_handles_labels()
ax.legend(handles, labels, title='Session / HRV Levels')
plt.show()
plt.close()
```
