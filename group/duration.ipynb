# Duration

Per-session response-duration breakdowns.

**Reads:** `data/group_results/ (10-participant summaries)`  
**Shared code:** `import mms` (loaders in `mms.io`, metrics in `mms.hrv` / `mms.stats`).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')

OUTPUT = '../data/group_results'

# load duration data
duration_df = pd.read_csv(f'{OUTPUT}/Psychometric_Test_Duration_STD.csv')

# melt to long format
melted = duration_df.melt(id_vars='Participant', var_name='Session', value_name='answer_duration')

# average per participant
average_durations = melted.groupby('Participant')['answer_duration'].mean().reset_index()

print("Average Answer Duration for Each Participant Across Three Sessions")
print(average_durations)

# plot averages
plt.figure(figsize=(12, 6))
sns.barplot(data=average_durations, x='Participant', y='answer_duration', palette='viridis')
plt.title('Average Answer Duration for Each Participant Across Three Sessions')
plt.ylabel('Average Answer Duration (seconds)')
plt.xlabel('Participant')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
plt.close()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')

OUTPUT = '../data/group_results'

# load duration data
duration_df = pd.read_csv(f'{OUTPUT}/Psychometric_Test_Duration_STD.csv')

# pivot by session
average_durations = duration_df.set_index('Participant')
average_durations.columns = ['Session 01', 'Session 02', 'Session 03']

print("Average durations (seconds):")
print(average_durations)

# plot averages
fig, ax = plt.subplots(figsize=(15, 8))
sns.lineplot(data=average_durations.T, markers=True, dashes=False, ax=ax)

# reference lines
ax.axhline(y=300, color='blue', linestyle=':', label='Normal')
ax.axhline(y=600, color='purple', linestyle=':', label='Moderate Duration')
ax.axhline(y=900, color='green', linestyle=':', label='High Duration')

# merge legend handles
handles, labels = ax.get_legend_handles_labels()
ax.legend(handles, labels, title='Participant / Duration Levels')

plt.title('Average Answer Duration for Each Participant Across Three Sessions')
plt.ylabel('Average Answer Duration (seconds)')
plt.xlabel('Session')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
plt.close()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')

OUTPUT = '../data/group_results'

# load duration data
duration_df = pd.read_csv(f'{OUTPUT}/Psychometric_Test_Duration_STD.csv')

# pivot by session
std_durations = duration_df.set_index('Participant')
std_durations.columns = ['Session 01', 'Session 02', 'Session 03']

print("Standard deviations (seconds):")
print(std_durations)

# transpose for plotting
std_durations = std_durations.T

# plot std deviations
fig, ax = plt.subplots(figsize=(15, 8))
sns.lineplot(data=std_durations, markers=True, dashes=False, ax=ax)

# reference lines
ax.axhline(y=300, color='blue', linestyle=':', label='Normal')
ax.axhline(y=600, color='purple', linestyle=':', label='Moderate Duration')
ax.axhline(y=900, color='green', linestyle=':', label='High Duration')

# merge legend handles
handles, labels = ax.get_legend_handles_labels()
ax.legend(handles, labels, title='Participant / Duration Levels')

plt.title('Standard Deviation of Answer Duration for Each Participant Across Three Sessions')
plt.ylabel('Standard Deviation of Answer Duration (seconds)')
plt.xlabel('Session')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
plt.close()
```
