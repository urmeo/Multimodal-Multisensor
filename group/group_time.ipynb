# Response-duration SD patterns

Ten participant summaries, three repeated visits. Historical estimator provenance is incomplete; source values remain unchanged, including documented exact duplicates. Metrics retain their separate units and do not measure anxiety, habituation or clinical severity. Clustering standardizes the three session SD values and uses three clusters when enough distinct complete rows exist. This is exploratory grouping of duration variability, not participant stress classes.

Data: `MMS_DATA_ROOT` or the checkout dataset. Exports: optional separate `MMS_OUTPUT_ROOT`.


```python
import matplotlib.pyplot as plt
from mms import exploration as ex
duration = ex.group_frames(["Psychometric_Test_Duration_STD"])
ex.plot_group(duration)
ex.plot_group(duration, distribution=True)
```

```python
frame = duration["Psychometric_Test_Duration_STD"]
projection, silhouette = ex.cluster_features(frame, n_clusters=3)
print("In-sample silhouette:", silhouette)
plt.scatter(projection.PC1, projection.PC2, c=projection.Cluster)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("Exploratory three-session duration-SD clusters, k=3")
plt.show()
plt.close()
```
