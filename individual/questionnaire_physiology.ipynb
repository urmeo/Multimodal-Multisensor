# Raw responses and descriptive physiology

All 88 item identities must occur exactly once. Stored answer codes lack verified option/reversal keys, so HADS subscale totals and universal STAI/FQ categories are unavailable. No radar normalization or clinical severity labels. Cross-clock physiology joins require acquisition timezone metadata.

Data: `MMS_DATA_ROOT` or the checkout dataset. Exports: optional separate `MMS_OUTPUT_ROOT`.


```python
import matplotlib.pyplot as plt
from mms import exploration as ex
SOURCE = 'individual'
heart = ex.recordings("hr", source=SOURCE)
hr_table = ex.hr_summary(heart)
print(hr_table)
intervals = ex.recordings("ibi", source=SOURCE)
interval_table = ex.interval_summary(intervals)
print(interval_table)
eyes = ex.recordings("eye", source=SOURCE)
eye_table = ex.eye_summary(eyes)
print(eye_table)
features = ex.aligned_questions(source=SOURCE)
if not features.empty:
    print(features.groupby(["Session", "Type", "Channel"])[["mean_hr_bpm", "interval_rmsdiff_ms", "mean_pupil_source_units"]].mean())
```

```python
durations, raw_codes = ex.response_tables(source=SOURCE)
print(durations)
print(raw_codes)
```

```python
for kind, part in raw_codes.groupby("Type"):
    plt.plot(part.Session, part.mean_raw_code, "o-", label=kind)
plt.title("Mean raw codes, separate instruments; not keyed scale scores")
plt.ylabel("Mean stored answer code")
plt.legend()
plt.show()
plt.close()
```
