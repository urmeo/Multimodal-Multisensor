# HADS item intervals and eye samples

This unidentified source stream is separate from the case-study recording. Pupil quality >=0.5. Legacy low-movement runs use step <0.01 and gaps <=0.1s. Closure candidates require both openness values <=1, quality >=0.5, observed opening boundaries, duration 0.05–0.5s and gaps <=0.1s. These settings are exploratory; no anxiety or validated blink inference. Historical raw codes lack verified reversal keys. No summed HADS scale or clinical category is inferred.

Data: `MMS_DATA_ROOT` or the checkout dataset. Exports: optional separate `MMS_OUTPUT_ROOT`.


```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mms import exploration as ex, hrv, io, stats
SOURCE = 'individual'
eyes = ex.recordings("eye", source=SOURCE)
eye_table = ex.eye_summary(eyes)
print(eye_table)
print(ex.baseline_changes(eye_table, ["Mean pupil (source units)", "Closure candidates/min"]))
```

```python
features = ex.aligned_questions(source=SOURCE)
if not features.empty:
    print(features.groupby(["Session", "Type", "Channel"])[["mean_hr_bpm", "interval_rmsdiff_ms", "mean_pupil_source_units"]].mean())
if not features.empty:
    hads = features.loc[features.Type.eq("HADS")]
    print(hads.groupby(["Session", "Channel"])[["n_pupil", "mean_pupil_source_units"]].agg(["count", "mean"]))
    ex.export_frame(hads, "individual/derived/hads_eye_item_features.csv")
```
