# Within-question response durations

Question start-to-answer differences are validated against Time(s). Durations and SDs are seconds; repeated items are not independent participants.

Data: `MMS_DATA_ROOT` or the checkout dataset. Exports: optional separate `MMS_OUTPUT_ROOT`.


```python
import matplotlib.pyplot as plt
from mms import exploration as ex
SOURCE = 'individual'
durations, raw_codes = ex.response_tables(source=SOURCE)
print(durations)
print(raw_codes)
```

```python
for kind, part in durations.groupby("Type"):
    plt.plot(part.Session, part.mean_response_s, "o-", label=kind)
plt.ylabel("Mean response duration (s)")
plt.legend()
plt.show()
plt.close()
```
