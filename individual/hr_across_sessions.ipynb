# HR and interval recordings

Channels remain separate. One available baseline is reused descriptively, not treated as a visit-specific baseline. No clinical HR/interval cutoffs. Question alignment requires acquisition timezone metadata.

Data: `MMS_DATA_ROOT` or the checkout dataset. Exports: optional separate `MMS_OUTPUT_ROOT`.


```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mms import exploration as ex, hrv, io, stats
SOURCE = 'individual'
heart = ex.recordings("hr", source=SOURCE)
hr_table = ex.hr_summary(heart)
print(hr_table)
ex.plot_summary(hr_table, "Mean HR (bpm)")
```

```python
intervals = ex.recordings("ibi", source=SOURCE)
interval_table = ex.interval_summary(intervals)
print(interval_table)
ex.plot_summary(interval_table, "Adjacent-sample RMS difference (ms)")
```

```python
features = ex.aligned_questions(source=SOURCE)
if not features.empty:
    print(features.groupby(["Session", "Type", "Channel"])[["mean_hr_bpm", "interval_rmsdiff_ms", "mean_pupil_source_units"]].mean())
```
