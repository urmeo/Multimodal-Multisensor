# Exploratory item-level associations

Separate recordings are never correlated by row index. Question containment requires explicit clock metadata. Instrument/channel descriptions are separate; Spearman coefficients and finite pair counts omit inferential p values for repeated within-person observations. Items and visits are repeated observations. Raw answer codes lack verified scale keys. Least squares is in-sample description, without diagnostic or predictive claims.

Data: `MMS_DATA_ROOT` or the checkout dataset. Exports: optional separate `MMS_OUTPUT_ROOT`.


```python
import matplotlib.pyplot as plt
from mms import exploration as ex
SOURCE = 'individual'
heart = ex.recordings("hr", source=SOURCE)
hr_table = ex.hr_summary(heart)
print(hr_table)
ex.plot_traces(heart, "hr")
```

```python
features = ex.aligned_questions(source=SOURCE)
if not features.empty:
    print(features.groupby(["Session", "Type", "Channel"])[["mean_hr_bpm", "interval_rmsdiff_ms", "mean_pupil_source_units"]].mean())
if not features.empty:
    for (kind, channel), part in features.groupby(["Type", "Channel"]):
        plt.scatter(part["response_s"], part["mean_hr_bpm"], s=12)
        plt.title(f"{kind}, channel {channel}: within-item descriptive association")
        plt.xlabel("Response duration (s)")
        plt.ylabel("Mean HR (bpm)")
        plt.show()
        plt.close()
```

```python
for (kind, channel), result in ex.question_correlations(features).items():
    print(f"{kind}, channel {channel}: repeated items; exploratory association only.")
    ex.plot_correlations(result)
```

```python
if not features.empty:
    print(features.groupby(["Type", "Channel"])[["response_s", "mean_hr_bpm", "interval_rmsdiff_ms", "mean_pupil_source_units"]].describe())
```
