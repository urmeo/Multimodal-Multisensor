# Interval-sample variability

These repeated interval samples are not established NN beat events. Channels remain separate; invalid samples break differences. The 30-sample window contains at most 29 internal pairs. Spectra use recorded elapsed time and have no clinical LF/HF interpretation.

Data: `MMS_DATA_ROOT` or the checkout dataset. Exports: optional separate `MMS_OUTPUT_ROOT`.


```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mms import exploration as ex, hrv, io, stats
SOURCE = 'case-study'
intervals = ex.recordings("ibi", source=SOURCE)
interval_table = ex.interval_summary(intervals)
print(interval_table)
ex.plot_traces(intervals, "ibi")
features = ex.aligned_questions(source=SOURCE)
if not features.empty:
    print(features.groupby(["Session", "Type", "Channel"])[["mean_hr_bpm", "interval_rmsdiff_ms", "mean_pupil_source_units"]].mean())
```

```python
spectra = ex.interval_spectra(intervals)
fig, axes = plt.subplots(max(1, len(spectra)), 1, figsize=(9, 2.5 * max(1, len(spectra))), squeeze=False)
for ax, (label, spectrum) in zip(axes.flat, spectra.items()):
    ax.plot(spectrum["Frequency (Hz)"], spectrum["Normalized sample power"])
    ax.set_title(label)
    ax.set_xlabel("Frequency (Hz)")
    ax.set_ylabel("Normalized sample power")
fig.suptitle("Recorded interval-sample spectra; no NN-beat or autonomic interpretation")
fig.tight_layout()
plt.show()
plt.close(fig)
```
