# Interval-sample summaries

Separate channels; invalid positions break adjacency. Values describe interval samples, not verified NN beats or health categories.

Data: `MMS_DATA_ROOT` or the checkout dataset. Exports: optional separate `MMS_OUTPUT_ROOT`.


```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mms import exploration as ex, hrv, io, stats
SOURCE = 'case-study'
intervals = ex.recordings("ibi", source=SOURCE)
interval_table = ex.interval_summary(intervals)
print(interval_table)
ex.plot_summary(interval_table, "Sample SD (ms)")
print(ex.baseline_changes(interval_table, ["Sample SD (ms)", "Adjacent-sample RMS difference (ms)"]))
```

```python
for label, frame in intervals.items():
    windows = hrv.hrv_over_time(frame, window_s=30, step_s=30)
    print(label, "30-second windows:", windows[["iSensor", "n_samples", "n_pairs", "sdnn", "rmssd"]].describe())
```
