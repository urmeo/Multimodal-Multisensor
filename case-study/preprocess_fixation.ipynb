# Build scratch low-movement events

Retain all eye rows and quality gaps. Store event durations once in a separate event table. Source CSVs stay immutable; set a separate MMS_OUTPUT_ROOT.

Data: `MMS_DATA_ROOT` or the checkout dataset. Exports: optional separate `MMS_OUTPUT_ROOT`.


```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mms import exploration as ex, hrv, io, stats
SOURCE = 'case-study'
for session in ex.SESSIONS:
    frame = io.load_raw("sed", session, source=SOURCE, data_root=ex.data_root())
    samples, events = ex.derive_eye(frame)
    suffix = "" if session is None else f"_{session:02d}"
    ex.export_frame(samples, f"{SOURCE}/derived/eye_samples{suffix}.csv")
    ex.export_frame(events, f"{SOURCE}/derived/low_movement_events{suffix}.csv")
```
