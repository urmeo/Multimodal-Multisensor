# Preprocess Raw To Csv

Convert semicolon-delimited raw sensor exports (.txt) into tidy processed CSVs.

**Reads:** `data/case-study/ (one participant, all sessions)`  
**Shared code:** `import mms` (loaders in `mms.io`, metrics in `mms.hrv` / `mms.stats`).

```python
import pandas as pd
import os
```

```python
RAW = '../data/case-study/raw'
PROCESSED = '../data/case-study/processed'

base_names = [
    'hr', 'hr_01', 'hr_02', 'hr_03',
    'sed', 'sed_01', 'sed_02', 'sed_03',
    'ibi', 'ibi_01', 'ibi_02', 'ibi_03',
]

for name in base_names:
    txt_path = os.path.join(RAW, f'{name}.txt')
    csv_path = os.path.join(PROCESSED, f'{name}.csv')
    df = pd.read_csv(txt_path, delimiter=';')
    df.to_csv(csv_path, index=False)
    print(f'{txt_path} -> {csv_path}  ({len(df)} rows)')
```
