# 2 Fixation

Eye-tracking fixation durations per session and per question type.

**Reads:** `data/case-study/ (one participant, all sessions)`  
**Shared code:** `import mms` (loaders in `mms.io`, metrics in `mms.hrv` / `mms.stats`).

```python
import pandas as pd

DATA = '../data/case-study/processed'

# load data
df1 = pd.read_csv(f'{DATA}/sed_fix_01.csv')
df2 = pd.read_csv(f'{DATA}/sed_fix_02.csv')
df3 = pd.read_csv(f'{DATA}/sed_fix_03.csv')
baseline_df = pd.read_csv(f'{DATA}/sed_fix.csv')

# avg fixation duration
def calculate_average_fixation_duration(df):
    df = df.copy()
    df['fixation_start'] = df['fixation'].diff().fillna(0) == 1
    df['fixation_end'] = df['fixation'].diff().fillna(0) == -1

    fixation_starts = df[df['fixation_start']].index
    fixation_ends = df[df['fixation_end']].index

    # validate pairs
    if len(fixation_starts) > len(fixation_ends):
        fixation_starts = fixation_starts[:len(fixation_ends)]
    elif len(fixation_ends) > len(fixation_starts):
        fixation_ends = fixation_ends[:len(fixation_starts)]

    # fixation durations
    fixation_durations = df.loc[fixation_ends, 'reltime'].values - df.loc[fixation_starts, 'reltime'].values
    fixation_durations = fixation_durations[~pd.isnull(fixation_durations)]
    average_duration = fixation_durations.mean() if len(fixation_durations) > 0 else 0
    return average_duration

# avg per session
baseline_avg_fixation = calculate_average_fixation_duration(baseline_df)
avg_fixation_01 = calculate_average_fixation_duration(df1)
avg_fixation_02 = calculate_average_fixation_duration(df2)
avg_fixation_03 = calculate_average_fixation_duration(df3)

# calculate difference
fixation_difference_01 = avg_fixation_01 - baseline_avg_fixation
fixation_difference_02 = avg_fixation_02 - baseline_avg_fixation
fixation_difference_03 = avg_fixation_03 - baseline_avg_fixation

# display results
print(f'Baseline Average Fixation Duration: {baseline_avg_fixation:.2f} seconds')
print(f'Session 1 Average Fixation Duration: {avg_fixation_01:.2f} seconds')
print(f'Difference from Baseline in Session 1: {fixation_difference_01:.2f} seconds')
print(f'Session 2 Average Fixation Duration: {avg_fixation_02:.2f} seconds')
print(f'Difference from Baseline in Session 2: {fixation_difference_02:.2f} seconds')
print(f'Session 3 Average Fixation Duration: {avg_fixation_03:.2f} seconds')
print(f'Difference from Baseline in Session 3: {fixation_difference_03:.2f} seconds')
```

```python
import pandas as pd
import numpy as np

DATA = '../data/case-study/processed'

# load data
data = pd.read_csv(f'{DATA}/sed_01.csv')

# fixation duration calc
def calculate_fixation_duration(df, velocity_threshold=0.1):
    df = df.copy()
    # gaze velocity
    df['gaze_velocity'] = np.sqrt(df['gazeDir.x'].diff()**2 +
                                  df['gazeDir.y'].diff()**2 +
                                  df['gazeDir.z'].diff()**2) / df['reltime'].diff()

    # identify fixations
    df['is_fixation'] = df['gaze_velocity'] < velocity_threshold

    # group fixations
    df['fixation_id'] = (df['is_fixation'] != df['is_fixation'].shift()).cumsum()

    # fixation durations
    g = df[df['is_fixation']].groupby('fixation_id')['reltime']
    fixation_durations = g.max() - g.min()

    # filter + average
    fixation_durations = fixation_durations[fixation_durations > 0]
    average_fixation_duration = fixation_durations.mean()

    return average_fixation_duration

# average fixation
average_fixation_duration = calculate_fixation_duration(data)

# convert to ms
average_fixation_duration_ms = average_fixation_duration * 1000

# print result
print(f"The average fixation duration is {average_fixation_duration_ms:.2f} milliseconds")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

DATA = '../data/case-study/processed'

baseline_data = pd.read_csv(f'{DATA}/sed.csv')
session_01_data = pd.read_csv(f'{DATA}/sed_01.csv')
session_02_data = pd.read_csv(f'{DATA}/sed_02.csv')
session_03_data = pd.read_csv(f'{DATA}/sed_03.csv')

# calculate saccade
def calculate_saccade_frequency(df, velocity_threshold=0.1):
    df = df.copy()
    # gaze velocity
    df['gaze_velocity'] = np.sqrt(df['gazeDir.x'].diff()**2 +
                                  df['gazeDir.y'].diff()**2 +
                                  df['gazeDir.z'].diff()**2) / df['reltime'].diff()

    # identify saccades
    df['is_saccade'] = df['gaze_velocity'] > velocity_threshold

    # count saccade events
    total_time = df['reltime'].iloc[-1] - df['reltime'].iloc[0]
    saccade_onsets = df['is_saccade'] & ~df['is_saccade'].shift(fill_value=False)
    saccade_count = saccade_onsets.sum()
    saccade_frequency = saccade_count / total_time if total_time > 0 else 0.0

    return saccade_frequency

# calculate saccade frequency
baseline_saccade_frequency = calculate_saccade_frequency(baseline_data)
session_01_saccade_frequency = calculate_saccade_frequency(session_01_data)
session_02_saccade_frequency = calculate_saccade_frequency(session_02_data)
session_03_saccade_frequency = calculate_saccade_frequency(session_03_data)

# frequency threshold
saccade_frequency_threshold = 5

# check saccade anxiety
increased_saccade_frequency_baseline = baseline_saccade_frequency > saccade_frequency_threshold
increased_saccade_frequency_01 = session_01_saccade_frequency > saccade_frequency_threshold
increased_saccade_frequency_02 = session_02_saccade_frequency > saccade_frequency_threshold
increased_saccade_frequency_03 = session_03_saccade_frequency > saccade_frequency_threshold

# display results
print(f'Baseline Saccade Frequency: {baseline_saccade_frequency:.2f} saccades/second - {"Increased" if increased_saccade_frequency_baseline else "Normal"}')
print(f'Session 1 Saccade Frequency: {session_01_saccade_frequency:.2f} saccades/second - {"Increased" if increased_saccade_frequency_01 else "Normal"}')
print(f'Difference from Baseline in Session 1: {session_01_saccade_frequency - baseline_saccade_frequency:.2f} saccades/second')
print(f'Session 2 Saccade Frequency: {session_02_saccade_frequency:.2f} saccades/second - {"Increased" if increased_saccade_frequency_02 else "Normal"}')
print(f'Difference from Baseline in Session 2: {session_02_saccade_frequency - baseline_saccade_frequency:.2f} saccades/second')
print(f'Session 3 Saccade Frequency: {session_03_saccade_frequency:.2f} saccades/second - {"Increased" if increased_saccade_frequency_03 else "Normal"}')
print(f'Difference from Baseline in Session 3: {session_03_saccade_frequency - baseline_saccade_frequency:.2f} saccades/second')

# plot saccade frequency
saccade_frequencies = [baseline_saccade_frequency, session_01_saccade_frequency, session_02_saccade_frequency, session_03_saccade_frequency]
labels = ['Baseline', 'Session 01', 'Session 02', 'Session 03']

plt.figure(figsize=(8, 6))
plt.bar(labels, saccade_frequencies, color=['blue', 'orange', 'green', 'red'])
plt.axhline(saccade_frequency_threshold, color='r', linestyle='dashed', linewidth=1, label='Saccade Frequency Threshold (>5 saccades/second)')
plt.ylabel('Saccade Frequency (saccades/second)')
plt.title('Saccade Frequency by Dataset')
plt.legend()
plt.show()
plt.close()
```

```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

DATA = '../data/case-study/processed'

# load fixation data
sessions = {}
for s in [0, 1, 2, 3]:
    filename = f'{DATA}/sed_fix.csv' if s == 0 else f'{DATA}/sed_fix_{s:02d}.csv'
    df = pd.read_csv(filename)
    label = 'Baseline' if s == 0 else f'Session {s:02d}'
    durations = df[df['fixation'] == True]['duration'].dropna()
    sessions[label] = durations

# descriptive stats
print("=== Fixation Duration Stats ===\n")
for name, dur in sessions.items():
    print(f"{name}: n={len(dur)}, mean={dur.mean():.4f}s, median={dur.median():.4f}s, SD={dur.std():.4f}s")

# Kruskal-Wallis
print("\n=== Kruskal-Wallis Test ===")
h, p = stats.kruskal(*sessions.values())
print(f"H={h:.3f}, p={p:.4f}")

# histograms
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
for ax, (name, dur) in zip(axes.flat, sessions.items()):
    ax.hist(dur, bins=50, edgecolor='black', alpha=0.7)
    ax.axvline(dur.mean(), color='red', linestyle='--', label=f'mean={dur.mean():.4f}s')
    ax.axvline(dur.median(), color='blue', linestyle=':', label=f'median={dur.median():.4f}s')
    ax.set_title(name)
    ax.set_xlabel('Fixation Duration (s)')
    ax.set_ylabel('Count')
    ax.legend(fontsize=8)

plt.suptitle('Fixation Duration Distributions')
plt.tight_layout()
plt.show()
plt.close()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

DATA = '../data/case-study/processed'
PSY = '../data/case-study/psychometric'

# load all modalities
hr_sessions = {}
ibi_sessions = {}
sed_sessions = {}
psy_sessions = {}

for s in [1, 2, 3]:
    hr = pd.read_csv(f'{DATA}/hr_{s:02d}.csv')
    hr = hr[hr['confidence'] == 1.0]
    hr['datetime'] = pd.to_datetime(hr['datetime'], utc=True, errors='coerce').dt.tz_convert(None)
    hr_sessions[s] = hr
    
    ibi = pd.read_csv(f'{DATA}/ibi_{s:02d}.csv')
    ibi['datetime'] = pd.to_datetime(ibi['datetime'], utc=True, errors='coerce').dt.tz_convert(None)
    ibi_sessions[s] = ibi
    
    sed = pd.read_csv(f'{DATA}/sed_{s:02d}.csv')
    sed['datetime'] = pd.to_datetime(sed['datetime'], utc=True, errors='coerce').dt.tz_convert(None)
    sed_sessions[s] = sed
    
    psy = pd.read_csv(f'{PSY}/Psychometric_Test_Results_{s:02d}.csv')
    psy['Question Start Time'] = pd.to_datetime(psy['Question Start Time'], utc=True, errors='coerce').dt.tz_convert(None)
    psy['Question Answer Time'] = pd.to_datetime(psy['Question Answer Time'], utc=True, errors='coerce').dt.tz_convert(None)
    psy_sessions[s] = psy

# baseline stats
hr_bl = pd.read_csv(f'{DATA}/hr.csv')
hr_bl = hr_bl[hr_bl['confidence'] == 1.0]
bl_hr_mean, bl_hr_std = hr_bl['heart_rate'].mean(), hr_bl['heart_rate'].std()

ibi_bl = pd.read_csv(f'{DATA}/ibi.csv')
ibi_clean = ibi_bl['ibi'][(ibi_bl['ibi'] > 300) & (ibi_bl['ibi'] < 2000)]
bl_sdnn = ibi_clean.std()
bl_rmssd = np.sqrt(np.mean(np.diff(ibi_clean.values)**2))

sed_bl = pd.read_csv(f'{DATA}/sed.csv')
bl_pupil_mean = sed_bl[sed_bl['pupil'] > 0]['pupil'].mean()
bl_pupil_std = sed_bl[sed_bl['pupil'] > 0]['pupil'].std()

# per-question biometrics
rows = []
for s in [1, 2, 3]:
    psy = psy_sessions[s]
    for _, q in psy.iterrows():
        start, end = q['Question Start Time'], q['Question Answer Time']
        
        # HR during question
        hr_seg = hr_sessions[s][(hr_sessions[s]['datetime'] >= start) & (hr_sessions[s]['datetime'] <= end)]
        avg_hr = hr_seg['heart_rate'].mean() if len(hr_seg) > 0 else np.nan
        
        # IBI/HRV during question
        ibi_seg = ibi_sessions[s][(ibi_sessions[s]['datetime'] >= start) & (ibi_sessions[s]['datetime'] <= end)]
        ibi_vals = ibi_seg['ibi'][(ibi_seg['ibi'] > 300) & (ibi_seg['ibi'] < 2000)]
        sdnn = ibi_vals.std() if len(ibi_vals) > 2 else np.nan
        rmssd = np.sqrt(np.mean(np.diff(ibi_vals.values)**2)) if len(ibi_vals) > 2 else np.nan
        
        # pupil during question
        sed_seg = sed_sessions[s][(sed_sessions[s]['datetime'] >= start) & (sed_sessions[s]['datetime'] <= end)]
        pupil_seg = sed_seg[sed_seg['pupil'] > 0]['pupil']
        avg_pupil = pupil_seg.mean() if len(pupil_seg) > 0 else np.nan
        
        # extract question number
        test_str = str(q['Test'])
        q_num = ''.join(c for c in test_str.split('.')[0] if c.isdigit())
        label = f"S{s} {q['Type']} Q{q_num}" if q_num else f"S{s} {q['Type']}"
        
        rows.append({
            'label': label, 'session': s, 'type': q['Type'],
            'HR': avg_hr, 'SDNN': sdnn, 'RMSSD': rmssd, 'Pupil': avg_pupil
        })

df = pd.DataFrame(rows)

# z-score relative to baseline
df['HR_z'] = (df['HR'] - bl_hr_mean) / bl_hr_std if bl_hr_std > 0 else 0
df['SDNN_z'] = -(df['SDNN'] - bl_sdnn) / bl_sdnn if bl_sdnn > 0 else 0
df['RMSSD_z'] = -(df['RMSSD'] - bl_rmssd) / bl_rmssd if bl_rmssd > 0 else 0
df['Pupil_z'] = (df['Pupil'] - bl_pupil_mean) / bl_pupil_std if bl_pupil_std > 0 else 0

# composite stress index
z_cols = ['HR_z', 'SDNN_z', 'RMSSD_z', 'Pupil_z']
df['Stress Index'] = df[z_cols].mean(axis=1)

# heatmap matrix
heatmap_data = df.set_index('label')[z_cols + ['Stress Index']].dropna()
heatmap_data.columns = ['HR', 'SDNN\n(inverted)', 'RMSSD\n(inverted)', 'Pupil', 'Composite\nStress']

plt.figure(figsize=(10, max(8, len(heatmap_data) * 0.25)))
sns.heatmap(heatmap_data, cmap='RdYlBu_r', center=0, annot=True, fmt='.1f',
            linewidths=0.5, cbar_kws={'label': 'Z-score (higher = more stress)'})
plt.title('Multimodal Stress Response per Question\n(z-scored vs baseline, SDNN/RMSSD inverted)')
plt.tight_layout()
plt.show()
plt.close()

# top stressed questions
top = df.nlargest(10, 'Stress Index')[['label', 'type', 'session', 'Stress Index']]
print("=== Top 10 Highest Stress Questions ===\n")
print(top.to_string(index=False))

# average by question type
print("\n=== Average Stress Index by Question Type ===\n")
type_stress = df.groupby('type')['Stress Index'].agg(['mean', 'std', 'count'])
print(type_stress.sort_values('mean', ascending=False).round(3).to_string())
```
