# Eye samples and low-movement runs

Legacy source-vector displacement <0.01 per sample, gaze quality >=0.5, gap <=0.1s. One observation per run; boundary/gap-censored runs are reported separately. This is an unvalidated low-movement heuristic. Pupil quality >=0.5; units follow the source. Closure candidates use both openness values <=1, eye quality >=0.5, observed reopening, duration 0.05–0.5s and gap <=0.1s. Openness polarity is an exploratory assumption, not a validated blink detector.

Data: `MMS_DATA_ROOT` or the checkout dataset. Exports: optional separate `MMS_OUTPUT_ROOT`.


```python
from mms import exploration as ex
SOURCE = 'case-study'
eyes = ex.recordings("eye", source=SOURCE)
eye_table = ex.eye_summary(eyes)
print(eye_table)
```

```python
print(ex.plot_events(eyes))
```

```python
print(ex.plot_events(eyes, histogram=True))
```

```python
ex.plot_summary(eye_table, "Mean run duration (ms)")
```

```python
for label, frame in eyes.items():
    events = ex.low_movement_events(frame)
    complete = events.loc[~events.left_censored & ~events.right_censored]
    subset = complete.loc[complete.duration_s.between(0.15, 0.30)]
    print(label, "exploratory 150–300ms subset:", len(subset), "events; mean ms:", subset.duration_s.mean()*1000)
```

```python
ex.plot_summary(eye_table, "Complete low-movement runs")
```
