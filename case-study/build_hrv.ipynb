# Build scratch interval variability

Trailing 30 interval samples per channel, at most 29 internal differences. Source rows and invalid positions are retained. Set MMS_OUTPUT_ROOT outside the input data; historical HRV files are not overwritten.

Data: `MMS_DATA_ROOT` or the checkout dataset. Exports: optional separate `MMS_OUTPUT_ROOT`.


```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mms import exploration as ex, hrv, io, stats
SOURCE = 'case-study'
for session in ex.SESSIONS:
    frame = io.load_ibi(session, source=SOURCE, data_root=ex.data_root())
    derived = hrv.hrv_rolling(frame, window_beats=30)
    suffix = "" if session is None else f"_{session:02d}"
    ex.export_frame(derived, f"{SOURCE}/derived/interval_variability{suffix}.csv")
```
